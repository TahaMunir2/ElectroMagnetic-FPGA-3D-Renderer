# MVP2 FDTD → HDMI bring-up

Real-time 2D FDTD electromagnetic wave, rendered as a ray-marched 3D heightmap on **HDMI** (640×480).

**What you should see:** plug HDMI into the PYNQ-Z1 J11 port → a shaded 3D terrain whose hills are the live |E| (or |S|) wave radiating from the injected source, updating continuously.

Pipeline (all on the 25 MHz pixel clock): `CORDIC → FDTD solver → field magnitude → ping-pong s_mag → bridge → 52 heightmap BRAMs → ray_unit → rgb2dvi → HDMI`.

## AXI map
| Base | Block | Use |
|------|-------|-----|
| `0x4000_0000` | renderer camera | optional camera pose (defaults are fine) |
| `0x4120_0000` | gpio_ctrl | FDTD control |
| `0x4121_0000` | gpio_status | FDTD status / checksum |

`gpio_ctrl` CH1 = `{amplitude[31:16], phase_step[15:0]}`, CH2 = `{free_run[15], sample_req[14], mag_mode[13], solver_enable[12], source_addr[11:0]}`.

`gpio_status` CH1 = `checksum[31:0]`, CH2 = `{source_q313[31:16], bridge_busy[7], pp_frame_ready[6], pp_read_sel[5], source_latched[4], mag_busy[3], mag_done[2], source_valid[1], solver_done[0]}`.

```python
from pynq import Overlay, MMIO
import time

# Copy fdtd_hdmi.bit (renamed from fdtd_hdmi_bd_wrapper.bit) and the matching .hwh
# to the board with the SAME basename, e.g. fdtd_hdmi.bit / fdtd_hdmi.hwh
ol = Overlay('fdtd_hdmi.bit')
print('Overlay loaded')
print([k for k in ol.ip_dict.keys()])
```

```python
CTRL   = MMIO(0x41200000, 0x10000)
STATUS = MMIO(0x41220000, 0x10000)
CAMERA = MMIO(0x40000000, 0x1000)
SRC    = MMIO(0x41210000, 0x10000)   # extra source positions + enables

GPIO_CH1 = 0x0   # data register, channel 1
GPIO_CH2 = 0x8   # data register, channel 2

GRID = 64
def cell(x, y):
    return y * GRID + x

def q313(v):
    """float -> signed Q3.13 16-bit (1.0 = 8192)"""
    return int(round(v * 8192)) & 0xFFFF

def set_ctrl(phase_step, amplitude, source_addr, solver_enable, mag_mode,
             sample_req, free_run, height_ctl=2):
    """height_ctl: signed -16..+15 terrain-height scale. >0 amplifies (left shift),
       <0 attenuates (right shift), 0 = pass-through. Soft source makes |E| small,
       so a positive value (+2..+4) is usually needed to see relief.
       mag_mode: 0=|E| (rectified, spiky), 1=|S| Poynting, 2=raw signed Ey
       (smooth hills+valleys 'water ripple' view)."""
    ch1 = (q313(amplitude) << 16) | q313(phase_step)
    ch2 = (((mag_mode >> 1) & 1) << 21) | ((height_ctl & 0x1F) << 16) | ((free_run & 1) << 15) \
        | ((sample_req & 1) << 14) | ((mag_mode & 1) << 13) | ((solver_enable & 1) << 12) \
        | (source_addr & 0xFFF)
    CTRL.write(GPIO_CH1, ch1)
    CTRL.write(GPIO_CH2, ch2)

def set_height(height_ctl):
    """Re-write just the height scale, keeping current ch2 bits, for live tuning."""
    ch2 = CTRL.read(GPIO_CH2) & ~(0x1F << 16)
    ch2 |= (height_ctl & 0x1F) << 16
    CTRL.write(GPIO_CH2, ch2)

def read_status():
    chk = STATUS.read(GPIO_CH1)
    s   = STATUS.read(GPIO_CH2)
    return {
        'checksum'      : chk,
        'solver_done'   : (s >> 0) & 1,
        'source_valid'  : (s >> 1) & 1,
        'mag_done'      : (s >> 2) & 1,
        'mag_busy'      : (s >> 3) & 1,
        'source_latched': (s >> 4) & 1,
        'pp_read_sel'   : (s >> 5) & 1,
        'pp_frame_ready': (s >> 6) & 1,
        'bridge_busy'   : (s >> 7) & 1,
        'source_q313'   : (s >> 16) & 0xFFFF,
    }
print('helpers ready')


def set_sources(positions, enables=None):
    """Configure up to 4 coherent sources for interference.
    positions: list of (x,y) cells. positions[0] uses the gpio_ctrl source_addr
               (set via set_ctrl(source_addr=...)); positions[1..3] go here.
    enables  : optional list of bools; default = all given positions enabled.
    Example: set_sources([(20,32),(44,32)])  # classic two-source interference
    """
    addr = [ (p[1]*GRID + p[0]) & 0xFFF for p in positions ]
    n = len(addr)
    if enables is None: enables = [True]*n
    en = 0
    for i in range(n):
        if enables[i]: en |= (1 << i)
    a1 = addr[1] if n > 1 else 0
    a2 = addr[2] if n > 2 else 0
    a3 = addr[3] if n > 3 else 0
    SRC.write(GPIO_CH1, (en << 24) | (a2 << 12) | a1)
    SRC.write(GPIO_CH2, a3)
    # source 0 address lives in gpio_ctrl CH2[11:0]; update it too
    ch2 = CTRL.read(GPIO_CH2) & ~0xFFF
    ch2 |= addr[0] & 0xFFF
    CTRL.write(GPIO_CH2, ch2)
    print(f"sources: {positions}  enables=0b{en:04b}")
```

## 1. Start the simulation in free-run mode

Inject a sine source at the grid centre, enable the solver in **free-run** (auto-restart after every magnitude pass), and hold `sample_req` high so the CORDIC keeps advancing the source phase.

```python
set_ctrl(
    phase_step   = 0.02,           # source frequency (lower = longer wavelength = bigger rings)
    amplitude    = 0.3,            # source amplitude (Q3.13). Soft source -> safe to keep modest
    source_addr  = cell(32, 32),   # centre of the 64x64 grid
    solver_enable= 1,
    mag_mode     = 2,              # 0=|E| (spiky), 1=|S|, 2=raw signed Ey (smooth waves)
    sample_req   = 1,
    free_run     = 1,
    height_ctl   = 2,              # +2 = x4 terrain height; tune live with set_height(n)
)
print('FDTD running (soft source). Tune relief live, e.g.  set_height(3)  or  set_height(1)')
set_sources([(32, 32)])   # enable one source at the centre (required: set_ctrl alone does not)
```

## 1b. Multiple sources — interference
Place 2 coherent sources and watch the interference fringes. Move them apart/together to change the fringe spacing. Up to 4 sources are supported (`source 0` = `gpio_ctrl` address; sources 1–3 via `gpio_src` @ 0x41230000).

```python
# two coherent sources, symmetric about the centre -> classic interference pattern
set_sources([(24, 32), (40, 32)])
# widen/narrow the spacing live, e.g. set_sources([(16,32),(48,32)]) or add more:
# set_sources([(24,24),(40,24),(32,44)])   # three sources
```

## Verify multi-source (PS-side smoke test)
Cycles through 1, 2, and 3 enabled sources and checks the solver stays live (checksum advancing, ping-pong swapping) for each configuration. This confirms the source GPIO (`0x41230000`) reaches the solver and injection is being driven. The full per-cell injection correctness is proven in RTL by `sim/tb_msrc.sv`.

```python
import time
def alive(n=15, dt=0.03):
    """Return (unique_checksums, read_sel_flips) over n status reads."""
    seen=set(); flips=0; prev=None
    for _ in range(n):
        s=read_status(); seen.add(s['checksum'])
        if prev is not None and s['pp_read_sel']!=prev: flips+=1
        prev=s['pp_read_sel']; time.sleep(dt)
    return len(seen), flips

# make sure the solver is running (soft source, wave view)
set_ctrl(phase_step=0.01, amplitude=0.3, source_addr=cell(24,32),
         solver_enable=1, mag_mode=2, sample_req=1, free_run=1, height_ctl=2)

configs = {
    '1 source ':[(32,32)],
    '2 sources':[(24,32),(40,32)],
    '3 sources':[(24,24),(40,24),(32,44)],
}
ok=True
for name,pos in configs.items():
    set_sources(pos); time.sleep(0.1)
    u,f = alive()
    live = (u>1 and f>0)
    print(f'{name}: unique_checksums={u:2d}  read_sel_flips={f:2d}  -> {"LIVE" if live else "DEAD"}')
    ok = ok and live

# disabling all sources should make the field decay (PML) -> eventually flat;
# checksum still moves while it drains, so we only assert the enabled cases here.
assert ok, 'a source configuration did not keep the solver live'
print('\nPASS: multi-source configurations all drive a live, swapping pipeline.')
# leave it on the nice 2-source interference pattern
set_sources([(24,32),(40,32)])
```

## 2. Confirm the solver is alive

The checksum must keep changing (solver writing fields), `pp_frame_ready`/`pp_read_sel` should toggle as frames complete, and `bridge_busy` pulses once per display vblank as it copies the wave into the heightmap.

```python
seen = set()
read_sel_flips = 0
prev_sel = None
for i in range(20):
    st = read_status()
    seen.add(st['checksum'])
    if prev_sel is not None and st['pp_read_sel'] != prev_sel:
        read_sel_flips += 1
    prev_sel = st['pp_read_sel']
    print(f"chk=0x{st['checksum']:08x} done={st['solver_done']} mag_done={st['mag_done']} "
          f"read_sel={st['pp_read_sel']} frame_rdy={st['pp_frame_ready']} bridge={st['bridge_busy']} "
          f"src=0x{st['source_q313']:04x}")
    time.sleep(0.05)

print()
print('unique checksums seen :', len(seen), '(should be > 1 => solver active)')
print('read_sel flips        :', read_sel_flips, '(> 0 => ping-pong swapping)')
assert len(seen) > 1, 'checksum frozen — solver not running'
print('PASS: solver + ping-pong are live.')
```

## 3. Look at the HDMI output

There is no framebuffer to read from the PS — the renderer streams straight to HDMI. **Check the monitor**: you should see a 3D terrain with concentric ripples expanding from the centre, damped at the edges by the PML boundary.

Try tuning the source live (re-run with different values):
* `phase_step` — wave frequency.
* `amplitude` — wave height.
* `mag_mode=1` — switch to |S| (Poynting) magnitude.
* `source_addr = cell(20, 40)` — move the source off-centre.

## 4. (Optional) camera pose

The renderer boots with a fixed default camera, so HDMI works without touching `0x4000_0000`. The `camera_ctrl_axi` block accepts a new origin/basis written to its shadow registers followed by a commit; it is applied at the next frame boundary (vsync) for tear-free camera moves. See `rtl/renderer/camera_ctrl_axi.sv` for the exact register offsets if you want to orbit the camera from the PS.
